#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
library(tidyverse)

marches_tidy <- marches |>
  # keeps only the columns needed for the research question
  # outcome (tourney_wins) and the predictors (seed, total_wins_pct)
  select(school, year, seed, total_wins_pct, tourney_wins, bid) |>
  # removes the 1983 seed = 0 rows, which are a special opening-round code, not a real seed
  filter(seed != 0) |>
  # drops rows with missing values in the variables we use
  drop_na() |>
  # treast bid as a categorical variable (factor)
  mutate(bid = as_factor(bid))

# shows the first 6 rows of the tidied data set that we just created
head(marches_tidy)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
marches_summary <- marches_tidy |>
#keep only the numeric columns 
#can only use (where(is.numeric)) to automatically select all numeric columns
select(year, seed, total_wins_pct, tourney_wins) |>
pivot_longer (cols = everything(), 
              names_to = "variable",
              values_to = "value") |>

group_by(variable) |>
summarize(mean = mean(value, na.rm = TRUE),
          min = min(value, na.rm = TRUE), 
          max = max(value, na.rm = TRUE))

marches_summary 
                  

             
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
marches_missing <- marches |>
    summarize(across(everything(), ~sum(is.na(.)))) |> 
    pivot_longer(cols = everything(), names_to = "column", values_to = "n_missing")

marches_missing

#note that we cannot use marches_tidy as we used "drop_na" before so if we were to sum up the NA's, it would be all 0
```

#### Visualization 1: Seed vs. Tournament Wins

This scatterplot shows the relationship between a team's seed and its number of corresponding tournament wins. The trend shows a strong negative relationship between the two variables: teams with low seed numbers (the strongest teams) win up 6 games, while teams with higher seed numbers rarely win more than 2 games. This suggests seed is a relative predictor of tournament wins, though the spread at each seed shows the prediction will not be exact.


```r
marches_seed_plot <- ggplot(marches_tidy, aes(x = seed, y = tourney_wins)) +
    geom_point(color = "steelblue1", alpha = 0.5) +
    labs(x = "Tournament Seed", y = "Number of Tournament Wins", title = "Relationship Between Tournament Seed and Tournament Wins") +
    theme(text = element_text(size = 17))
marches_seed_plot
```

#### Visualization 2: Total Win Percentages vs. Tournament Wins

This scatterplot shows the relationship between total win percentages against tournament wins. The trend shows a strong positive relationship between the two variables: teams with higher win percentages tend to win more tournament games; however, the trend is a bit noisy, as there are many teams with zero wins across all the win percentages (x-axis), and there is one outlier with a low win percentage and three wins. Win percentage is therefore a useful but imperfect predictor on its own, but when combined with seed, it can help the model.

```r
marches_pct_plot <- ggplot(marches_tidy, aes(x = total_wins_pct, y = tourney_wins)) +
    geom_point(color = "seagreen3", alpha = 0.5) +
    labs(x = "Regular-Season Win Percentage", y = "Number of Tournament Wins", title = "Relationship Between Regular-Season Win Percentage and Tournament Wins") +
    theme(text = element_text(size = 17))
marches_pct_plot
```

I plan to use K-nearest neighbour (k-NN) regression to predict the number of tournament wins (tourney_wins) from both a team's seed and total win percentage. Regression is appropriate because the outcome that we are trying to predict is a numeric value (if it were a category, we would use classification), and my scatterplots show that both predictors are related to it (negatively for seed, and positively for win percentage). k-NN also makes no assumptions about the shape of the relationship. Its weakness is that it is sensitive to predictor scales, so the predictors must first be standardized; it cannot predict outside the range of the given training data and it also gives no interpretable coefficients. Lastly, since tournament wins are whole numbers with many zeroes, its averaged predictions may not be whole numbers. 

Initially, I will first split the data into 75% training and 25% testing using initial_split, stratified by the dataframe of tourney_wins, with a set seed for reproducibility. The test set will remain untouched until the very end. On the training set, I will standardize the predictors with a recipe (step_center along with step_scale) and use 5-fold cross-validation to choose the proper K value, picking the K with the lowest cross-validation RMSE. I will not create a separate validation set, because cross-validation plays that role. I will then refit the model on the whole training set using the newly picked and best K.

I will assess performance by calculating the RMSPE on the test set. RMSPE is appropriate because it measures, in the same units as the outcome of tourney_wins, how far the predictions typically are from the true values of the data the model has not seen yet. Lower values equate to better predictions
